```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('G:\project\Thiranex\Road_Accident_Analysis\data\Cleaned_Road_Accident_Data.csv')

df.head()
```

### Statistical Summary

```python
df.describe()
```

```python
df.describe(include='object')
```

### Accident Severity Analysis

```python
sns.countplot(x='Accident_Severity', data=df)
```

### Day-of-Week Analysis

```python
plt.figure(figsize=(10, 6))

sns.countplot(x='Day_of_Week', data=df, order=df['Day_of_Week'].value_counts().index)
```

### Hourly Trend Analysis

```python
sns.histplot(df['Hour'], bins=24)
```

### Monthly Trend Analysis

```python
sns.countplot(x='Month', data=df)
```

### Weather Analysis

```python
weather = df['Weather_Conditions'].value_counts().head(10)

sns.barplot(x=weather.values, y=weather.index)
```

### Road Surfce Analysis

```python
road = (
    df['Road_Surface_Conditions']
    .value_counts()
)

sns.barplot(
    x=road.values,
    y=road.index
)
```

### Urban vs Rural Anlysis

```python
sns.countplot(x='Urban_or_Rural_Area', data=df)
```

### Vehicle Type Analysis

```python
vehicle = df['Vehicle_Type'].value_counts().head(10)

sns.barplot(x=vehicle.values, y=vehicle.index)
```

### Casualty Analysis

```python
sns.histplot(df['Number_of_Casualties'], bins=20)
```

### Correlation Analysis

```python
numeric_df = df.select_dtypes(include='int64')
corr = numeric_df.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

### Relationship Analysis

```python
sns.scatterplot(x='Speed_limit', y='Number_of_Casualties', data=df)
```

### Vehicles vs Casulaties

```python
sns.scatterplot(x='Number_of_Vehicles', y='Number_of_Casualties',hue='Accident_Severity', data=df)
```

### Geographic Analysis

```python
plt.figure(figsize=(10, 8))

plt.scatter(df['Longitude'], df['Latitude'], alpha=0.2)
plt.title('Accident Locations')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```
